## Introducción

En 1912, el personal encargado del embarque del Titanic utilizaba listas en papel para registrar a los pasajeros. En esta práctica, estudiaremos ese registro como un ejemplo de procesamiento manual de transacciones. Las listas cumplían una función operativa: dejar constancia de quiénes viajaban. El dataset reúne datos de los pasajeros, como su clase y puerto de embarque, e incorpora un dato conocido después del naufragio: si sobrevivieron.

### Diccionario de datos

El archivo `titanic.csv` contiene **891 registros y 12 columnas**. Cada fila representa a un pasajero.

| Columna | Tipo de dato | Descripción |
|---|---|---|
| `PassengerId` | Identificador entero | Código único asignado a cada pasajero en el dataset. |
| `Survived` | Categoría binaria | Resultado registrado después del naufragio. `0` = falleció, `1` = sobrevivió. |
| `Pclass` | Categoría ordinal | Clase del boleto. `1` = primera, `2` = segunda, `3` = tercera. |
| `Name` | Texto | Nombre del pasajero tal como aparece en el archivo. |
| `Sex` | Categoría | Sexo registrado. Los valores del archivo son `female` y `male`. |
| `Age` | Número | Edad en años. Puede incluir valores fraccionarios, por ejemplo, para bebés menores de un año. |
| `SibSp` | Conteo entero | Número de hermanos, hermanas y cónyuges del pasajero a bordo. |
| `Parch` | Conteo entero | Número de padres, madres, hijos e hijas del pasajero a bordo. |
| `Ticket` | Código de texto | Código del boleto. Puede incluir letras, números y signos, y repetirse en los registros de varios pasajeros. |
| `Fare` | Número decimal | Tarifa registrada en la fila del pasajero. |
| `Cabin` | Código de texto | Identificación de la cabina, cuando figura en el registro. |
| `Embarked` | Categoría | Puerto de embarque. `C` = Cherburgo, `Q` = Queenstown, `S` = Southampton. |

Presenta los resultados en tablas con encabezados claros e indica cómo trataste los valores faltantes cuando afecten un cálculo. Todos los resultados deben referirse a los pasajeros del archivo.

```python
import pandas as pd
df = pd.read_csv("titanic.csv", sep=',', header = 0, encoding = 'utf-8')
display(df.head())

```

```python
print("1. ¿Cuántos pasajeros hay en el dataset y cuántos PassengerId distintos aparecen? ¿Coinciden los dos resultados?")

cont = 0
for i in range(len(df)):
    cont += 1

id_distintos = len(df['PassengerId'].unique())

pasajeros_dataset = pd.DataFrame({
    'Numero de pasajeros': [cont],
    'Id distintos': [id_distintos],
})

display(pasajeros_dataset)

if cont == id_distintos:
    print("Coinciden los resultados")
else:
    print("No coinciden los resultados")


```

```python
print("2. ¿Cuántos pasajeros de cada clase aparecen en el archivo? ¿Qué clase tiene más pasajeros registrados?")

clase_pasajeros = df['Pclass'].agg(['value_counts']).rename(columns={'value_counts': 'Cantidad de pasajeros' }).rename_axis('Clase').sort_index()
display(clase_pasajeros)

mayor = df['Pclass'].value_counts().idxmax()

if mayor == 1:
    print("La primera clase tiene más pasajeros registrados")
if mayor == 2:
    print("La segunda clase tiene más pasajeros registrados")
if mayor == 3:
    print("La tercera clase tiene más pasajeros registrados")

```

```python
print("3. ¿Cuántos pasajeros embarcaron en cada puerto y cuántos tienen el puerto sin registrar?")

pasajeros_por_puerto = df['Embarked'].agg(['value_counts']).rename(columns={'value_counts':'Cantidad de pasajeros'}).rename_axis('Puerto').sort_index()
display(pasajeros_por_puerto)

pasajeros_sin_puerto = df['Embarked'].isnull().sum()
print("Cantidad de pasajeros sin puerto registrados: ", pasajeros_sin_puerto)
```

Aqui utilizamos el metodo del pandas value_count(), ya que solo evalua lo no nulos, y luego se usaron los metodos
 .isnull().sum() para contar lo nulos.

```python
print("4. ¿Cuál es la tarifa promedio, mínima y máxima registrada para cada clase?")
resultados_tarifa = df.groupby('Pclass')['Fare'].agg(['mean', 'min', 'max']).rename_axis('Clase').round(2).sort_index()
display(resultados_tarifa)

```

```python
print("5. ¿Cuántos pasajeros de cada clase embarcaron en cada puerto?")
class_embarked = df.groupby('Pclass')['Embarked'].agg({'value_counts'}).rename(columns={'value_counts':'Cantidad de pasajeros'}).sort_index()

class_embarked = class_embarked.rename_axis(['Clase', 'Puerto'])

display(class_embarked)
```

Debido al metodo value_counts() no se tomaron en cuenta las 2 personas que no tienen registro.

```python
print("6. ¿Cuántos pasajeros sobrevivieron y cuántos fallecieron en cada clase?")
sobrevivientes_por_clase = df.groupby('Pclass')['Survived'].agg(['value_counts']).rename(columns={'value_counts':'Cantidad de pasajeros'}).sort_index()
sobrevivientes_por_clase = sobrevivientes_por_clase.rename_axis(['Clase', 'Estado del ser humano'])
sobrevivientes_por_clase = sobrevivientes_por_clase.replace({0:'Fallecido', 1:'Sobreviviente'})

display(sobrevivientes_por_clase)


```

```python
print("7. ¿Cuántos pasajeros viajaban con cero, uno o dos o más familiares registrados a bordo?")
df['Familiares totales'] = df['SibSp'] + df['Parch']

familiares = df['Familiares totales'].agg(['value_counts']).rename(columns={'value_counts':'Cantidad de pasajeros'}).sort_index()
display(familiares)
```

```python
print("8. ¿Cuántos pasajeros tenían menos de 18 años, cuántos tenían 18 o más y cuántos tenían la edad sin registrar? Desglose el resultado por clase.")

def clasificar_edad(edad):
    if pd.isnull(edad):
        return 'Sin registro'
    elif edad < 18:
        return 'Menor de 18'
    elif edad >= 18:
        return 'Mayor o igual a 18'

df['Clase_edad'] = df['Age'].apply(clasificar_edad)
edad_clase = df.groupby('Pclass')['Clase_edad'].agg(['value_counts']).rename(columns={'value_counts':'Cantidad de pasajeros'}).sort_index()
display(edad_clase)
```

Se definio una funcion en la que se considero los valores nulos, luego se llamo a la funcion para clasificar los datos de la columna Age y luego lo guardamos en una nueva, y asi no se les dejo fuera del calculo final.

```python
print("9. ¿Cuántos códigos de boleto distintos hay? ¿Cuántos aparecen asociados a más de un pasajero y cuál aparece en más registros?")

codigos_distintos = df['Ticket'].nunique()
frecuencia_codigos = df['Ticket'].value_counts()
codigos_repetidos = frecuencia_codigos[frecuencia_codigos > 1]
codigo_mas_frecuente = frecuencia_codigos.idxmax()
mayor_frecuencia = frecuencia_codigos.max()

resumen_boletos = pd.DataFrame({
    'Indicador': [
        'Códigos distintos',
        'Códigos asociados a más de un pasajero',
        'Código más frecuente',
        'Mayor cantidad de apariciones'
    ],
    'Resultado': [
        codigos_distintos,
        len(codigos_repetidos),
        codigo_mas_frecuente,
        mayor_frecuencia
    ]
})

display(resumen_boletos)
```

```python
print("10. ¿En qué clase hay más registros sin cabina? Para cada clase, muestren cuántos registros tienen cabina, cuántos carecen de ella y cuántos hay en total")

sin_cabina = df[df['Cabin'].isnull()]
total_por_clase = df.groupby('Pclass').size()
sin_cabina_por_clase = sin_cabina.groupby('Pclass').size()
con_cabina_por_clase = total_por_clase - sin_cabina_por_clase

resumen_cabina = pd.DataFrame({
    'Clase': [1, 2, 3],
    'Con cabina': con_cabina_por_clase.values,
    'Sin cabina': sin_cabina_por_clase.values,
    'Total': total_por_clase.values
})

display(resumen_cabina)

    
```

Utilizamos metodos de pandas para recorrer el dataFrame y guardar las cabinas vacias, para luego respresentar en la tabla.
